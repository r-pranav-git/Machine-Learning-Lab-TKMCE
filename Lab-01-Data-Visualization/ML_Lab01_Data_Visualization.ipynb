**1) Programs to handle data using NumPy, Pandas and matplotlib/plotly/bokeh/seaborn for data visualization**

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset
df = pd.read_csv("Iris.csv")

# Display first 5 rows
print("First Five Rows:")
display(df.head())

# Shape of dataset
print("Shape of Dataset:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns)

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical Summary
print("\nStatistical Summary:")
display(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Set seaborn style
sns.set_style("whitegrid")

# Scatter Plot
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="SepalWidthCm",
    hue="Species"
)

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.legend(title="Species")

plt.show()
```

**2) Calculate the mean, median, and mode of a given numeric dataset (Iris dataset).**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Iris.csv")

# Remove the Id column
df = df.drop("Id", axis=1)

# Mean
print("Mean")
print(df.mean(numeric_only=True))

# Median
print("\nMedian")
print(df.median(numeric_only=True))

# Mode
print("\nMode")
print(df.mode(numeric_only=True))
```

**3) Create a histogram to visualize the distribution of sepal length across all Iris samples.**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("Iris.csv")

# Plot histogram
plt.figure(figsize=(8,5))
plt.hist(df["SepalLengthCm"], bins=10, color="skyblue", edgecolor="black")

# Add title and labels
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

**4)Create histograms of sepal width for each Iris species in a single plot, using different colors for each species.**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("Iris.csv")

# Histogram for each species
plt.figure(figsize=(8,5))

plt.hist(df[df["Species"]=="Iris-setosa"]["SepalWidthCm"],
         bins=10, alpha=0.5, color="blue", label="Setosa")

plt.hist(df[df["Species"]=="Iris-versicolor"]["SepalWidthCm"],
         bins=10, alpha=0.5, color="green", label="Versicolor")

plt.hist(df[df["Species"]=="Iris-virginica"]["SepalWidthCm"],
         bins=10, alpha=0.5, color="red", label="Virginica")

plt.title("Histogram of Sepal Width for Each Species")
plt.xlabel("Sepal Width (cm)")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

**5) Create a boxplot to compare sepal length across the three Iris species.**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("Iris.csv")

# Create boxplot
plt.figure(figsize=(8,5))

sns.boxplot(x="Species", y="SepalLengthCm", data=df)

plt.title("Sepal Length Comparison of Iris Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")

plt.show()
```

**6) Create a scatterplot to compare petal length and petal width, colored by species.**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("Iris.csv")

# Create scatter plot
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="PetalLengthCm",
    y="PetalWidthCm",
    hue="Species",
    data=df
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.show()
```
